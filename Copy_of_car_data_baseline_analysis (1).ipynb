```python
import pandas as pd
import numpy as np

# Creating messy dataset
data = {
    "brand": ["Toyota ", "Honda", "Ford", "Toyota", "Honda", "BMW", "Audi", None, "Ford", "BMW"],
    "mileage": ["20 kmpl", "18 kmpl", "15 kmpl", None, "22 kmpl", "12 kmpl", "14 kmpl", "16 kmpl", None, "13 kmpl"],
    "selling_price": [500000, 600000, None, 550000, 620000, 1500000, 1800000, 700000, 580000, None],
    "year": [2015, 2016, 2014, 2015, 2017, 2018, 2019, 2016, 2015, 2017]
}

df = pd.DataFrame(data)

# Save dataset
df.to_csv("car_data.csv", index=False)

df
```

```python
df = pd.read_csv("car_data.csv")

print("Shape:", df.shape)
df.head()
```

```python
print("\n--- INFO ---")
df.info()

print("\n--- DESCRIBE ---")
print(df.describe())

print("\n--- SHAPE ---")
print(df.shape)
```

Data Quality Issues:

1. Missing values in mileage and selling_price
2. Mileage column contains text (kmpl) instead of numeric
3. Brand column has inconsistent formatting (extra spaces, uppercase)
4. Duplicate rows may exist

```python
df = df.dropna(subset=["selling_price"])
```

```python
# Fill numeric column
df["year"] = df["year"].fillna(df["year"].median())

# Fill categorical columns
df["brand"] = df["brand"].fillna(df["brand"].mode()[0])
df["mileage"] = df["mileage"].fillna(df["mileage"].mode()[0])
```

```python
df["brand"] = df["brand"].str.strip().str.lower()
```

```python
df["mileage"] = df["mileage"].astype(str)
df["mileage"] = df["mileage"].str.extract(r'(\d+\.?\d*)')
df["mileage"] = pd.to_numeric(df["mileage"], errors='coerce')
```

```python
df = df.drop_duplicates()

print("Cleaned Shape:", df.shape)
```

```python
import numpy as np

mean_price = df["selling_price"].mean()

predictions = np.full(len(df), mean_price)

mae = np.mean(np.abs(df["selling_price"] - predictions))

print("Mean Selling Price:", mean_price)
print("Baseline MAE:", mae)
```
